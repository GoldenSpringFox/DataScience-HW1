# Synergy communities via Louvain

**Task 6.3-C** (`docs/devlog/6.3c-louvain-communities.md`). Replaces the NMF / symNMF / theme-label
exploration, archived at `notebooks/archive/communities_exploration.ipynb` — 40 soft topics that
each bundled ~22 distinct themes and needed four bespoke modules to interpret.

One graph, one algorithm, one picture: **§3 is a stats overview, §4 is the graph.**

| Step | Tool |
|---|---|
| Edge weights | `cooccurrence` colour-conditioned t-score, near-uniform deck-weighted |
| Sparsification | `sparsify_top_k_union` — each card's 15 strongest positive edges, union-symmetrized |
| Communities | `networkx...louvain_partitions` — the algorithm from class; yields every level of the recursive merge, so granularity comes free |
| Interactive graph | `ipysigma` — ForceAtlas2 in the browser on WebGL; strong edges pull cards together |

No igraph, no leidenalg, no NMF, no embeddings.

### Two decisions baked in

**Deck weighting is not optional.** The corpus is 13,207 decks across 1,001 commander slots, and
two commanders are a quarter of it (Krenko 2,007 = 15.2%, Kyler 1,235 = 9.4%) against a median slot
of ~8 decks. Uncorrected, their packages seed large same-colour blobs — visible in the graph as one
giant goblin cluster. `compute_near_uniform_weights` flattens each slot's contribution (a Krenko
deck gets 0.0040 against a median slot's 1.0), which cuts Krenko's signature capture from 93% to
72% and Kyler's from 83% to 39%. **The saved `tscore_color_global.npz` is unweighted** —
`cooccurrence.build_and_save` never passes deck weights — so §2 rebuilds the matrix and caches it.
§3's guard re-checks this every run.

**Colour bias is reported, not corrected.** Louvain on the plain and colour-conditioned matrices
give the same purity (~53% vs a ~20% null), as did NMF. The colour signal is in the data, not in
any algorithm — cards of one colour genuinely do appear in decks together. §3 measures it; the
right tail is where colour and mechanic honestly coincide (Devils really are red).

**gamma = 2.0**, not 1.0: the weighted graph is less blocky than the unweighted one, and at gamma=1
its largest community balloons to 2,762. Raising it to 2.0 gives 214 communities with the largest
at 868 and 114 package-sized.

## 1. Setup

```python
from pathlib import Path

import numpy as np
import pandas as pd
import networkx as nx
from networkx.algorithms import community as nxc
from scipy import sparse

from edhcut.analysis.communities import basic_land_mask, sparsify_top_k_union
from edhcut.analysis.cooccurrence import (
    KB_DEV_DIR,
    build_card_color_identities,
    build_color_identity_deck_counts,
    build_cooccurrence,
    compute_color_conditioned_tscore,
    load_card_index,
)
from edhcut.analysis.deck_weights import compute_near_uniform_weights
from edhcut.config import CONFIG
from edhcut.db import connect

TOP_K = 15        # per-card strongest positive edges kept (union-symmetrized)
RESOLUTION = 2.0  # Louvain gamma -- see header
SEED = 42
EDGE_THRESHOLD = 3.0  # minimum t-score for an edge to be *drawn* in section 4
```

## 2. Build

Weighted co-occurrence &rarr; colour-conditioned t-score &rarr; sparsified graph &rarr; Louvain.
The rebuild costs ~80s and is cached, so re-runs are instant.

```python
card_index = load_card_index()
CACHE = Path(KB_DEV_DIR) / "tscore_color_global_nearuniform.npz"

with connect(CONFIG.paths.db_path) as conn:
    basics = basic_land_mask(conn, card_index)
    color_identity = np.array(build_card_color_identities(conn, card_index), dtype=object)

    if CACHE.exists():
        tscore = sparse.load_npz(CACHE)
        total_weight = None
    else:
        print("building near-uniform-weighted co-occurrence (~80s, cached afterwards)...")
        result = build_cooccurrence(
            conn, card_index, slot_key=None,
            deck_slot_weights=compute_near_uniform_weights(conn),
        )
        tscore = compute_color_conditioned_tscore(
            result,
            build_card_color_identities(conn, card_index),
            build_color_identity_deck_counts(conn, slot_key=None),
        )
        sparse.save_npz(CACHE, tscore)
        total_weight = result.total_weight

    deck_total, slot_total = conn.execute(
        "SELECT COUNT(*), COUNT(DISTINCT slot_key) FROM decks"
    ).fetchone()

# Graph: positive t-score only, each card's TOP_K strongest edges, union-symmetrized.
adjacency, kept = sparsify_top_k_union(tscore, top_k=TOP_K, min_tscore=0.0, exclude=basics)
G = nx.from_scipy_sparse_array(adjacency, edge_attribute="weight")

levels = list(nxc.louvain_partitions(G, weight="weight", resolution=RESOLUTION, seed=SEED))


def labels_from(partition) -> np.ndarray:
    out = np.empty(G.number_of_nodes(), dtype=int)
    for cid, members in enumerate(sorted(partition, key=len, reverse=True)):
        out[list(members)] = cid
    return out


cards = card_index.loc[kept].reset_index(drop=True)[["oracle_id", "name", "deck_count"]].copy()
cards["node"] = cards.index                          # row order == graph node id
cards["community"] = labels_from(levels[0])          # fine -- package-sized
cards["coarse"] = labels_from(levels[-1])            # coarse -- for graph colouring
cards["strength"] = np.array([s for _, s in G.degree(weight="weight")]).round(2)
cards["color_identity"] = ["C" if c == "" else c for c in color_identity[kept]]
```

### Naming the communities automatically

`card_tags` (Scryfall Tagger, ~377k rows over a ~950-tag vocabulary) is a far better naming signal
than type line or mana cost, because it already encodes *mechanics* (`landfall`, `synergy-desert`,
`typal-goblin`) rather than surface features.

For each (community, tag): `share` = fraction of the community carrying the tag, `lift` = how much
more often than the graph-wide base rate. Score them by `share x log(1 + lift)` — `share` alone
picks vacuous tags every card has (`triggered-ability`), `lift` alone picks freak tags on 4 cards.

The failures are the useful part: **a community whose best tag has lift < 3 has no mechanical theme
at all.** That is exactly what the generic-staple communities look like, so the naming doubles as a
quality flag.

```python
MIN_TAG_CARDS = 4     # ignore tags carried by fewer than this many cards in the community
GENERIC_LIFT = 3.0    # best-tag lift below this => no real theme

with connect(CONFIG.paths.db_path) as conn:
    tags = pd.read_sql("SELECT oracle_id, tag FROM card_tags", conn)

tags = tags[tags["oracle_id"].isin(set(cards["oracle_id"]))].copy()
tags["community"] = tags["oracle_id"].map(dict(zip(cards["oracle_id"], cards["community"])))

_sizes = cards["community"].value_counts()
_tag_total = tags.groupby("tag")["oracle_id"].nunique()
tag_stats = (tags.groupby(["community", "tag"])["oracle_id"].nunique().rename("k").reset_index())
tag_stats["share"] = tag_stats["k"] / tag_stats["community"].map(_sizes)
tag_stats["lift"] = tag_stats["share"] / (tag_stats["tag"].map(_tag_total) / len(cards))
tag_stats["score"] = tag_stats["share"] * np.log1p(tag_stats["lift"])


def community_label(cid: int, n_tags: int = 2) -> str:
    """Auto-name a community from its most distinctive Scryfall tags."""
    sub = tag_stats[(tag_stats["community"] == cid) & (tag_stats["k"] >= MIN_TAG_CARDS)]
    sub = sub.nlargest(n_tags, "score")
    if sub.empty or sub["lift"].max() < GENERIC_LIFT:
        return "(generic - no distinctive tag)"
    return " / ".join(sub["tag"])


community_names = {cid: community_label(cid) for cid in _sizes.index}
cards["label"] = cards["community"].map(community_names)
```

## 3. Overview

```python
def overview() -> None:
    """Everything worth knowing about this build, in one block."""
    sizes = cards["community"].value_counts()
    degrees = np.array([d for _, d in G.degree()])
    weights = np.array([w for _, _, w in G.edges(data="weight")])
    rng = np.random.default_rng(SEED)

    print("CORPUS")
    print(f"  {deck_total:,} decks across {slot_total:,} commander slots")
    print(f"  near-uniform deck weighting applied"
          + (f" -> total weight {total_weight:,.0f}" if total_weight else " (cached matrix)"))
    print(f"  {len(card_index):,} cards in index, {int(basics.sum())} basic lands excluded")

    print("\nGRAPH  (positive t-score, top-{} union-symmetrized)".format(TOP_K))
    print(f"  {G.number_of_nodes():,} nodes  ({(~kept).sum():,} cards dropped: no surviving edge)")
    print(f"  {G.number_of_edges():,} edges")
    print(f"  degree      : median {int(np.median(degrees))}, max {degrees.max():,}")
    print(f"  edge t-score: median {np.median(weights):.2f}, p90 {np.percentile(weights, 90):.2f}, "
          f"max {weights.max():.1f}")

    print(f"\nLOUVAIN  (gamma={RESOLUTION}, seed={SEED})")
    for i, part in enumerate(levels):
        s = np.array(sorted((len(c) for c in part), reverse=True))
        print(f"  level {i}: {len(part):4d} communities  Q={nxc.modularity(G, part, weight='weight', resolution=RESOLUTION):.3f}"
              f"  largest {s[0]:5,}  median {int(np.median(s)):4d}"
              f"  package-sized(8-60) {int(((s >= 8) & (s <= 60)).sum()):4d}"
              f"  singletons {int((s == 1).sum())}")
    print(f"  working partition: level 0 ({cards['community'].nunique()} fine) "
          f"+ level {len(levels)-1} ({cards['coarse'].nunique()} coarse, for colouring)")

    # Colour purity vs a same-size random null.
    measured = cards.groupby("community").filter(lambda g: len(g) >= 8)
    purity = measured.groupby("community")["color_identity"].agg(
        lambda s: s.value_counts().iloc[0] / len(s))
    pool = cards["color_identity"].to_numpy()
    null = np.array([
        pd.Series(rng.choice(pool, size=int(n), replace=False)).value_counts().iloc[0] / n
        for n in np.repeat(measured.groupby("community").size().to_numpy(), 10)
    ])
    print("\nCOLOUR PURITY  (reported, not corrected -- it is in the data, not the algorithm)")
    print(f"  median {purity.median():.1%} across {len(purity)} communities >=8 cards"
          f"   (random null {np.median(null):.1%})")
    print(f"  >=70%: {(purity >= .7).sum()}   >=80%: {(purity >= .8).sum()}   "
          f">=90%: {(purity >= .9).sum()}   single-colour: {(purity >= .999).sum()}")

    # Guard: has any oversampled commander recaptured a community?
    print("\nSLOT DOMINATION GUARD  (share of a commander's signature cards landing in one community)")
    slots = {"Krenko": "68418069-f615-40ef-ae0d-764192acae00",
             "Kyler": "726cd041-5d0b-436c-bced-9335f56c0b0d",
             "Yenna": "0adede77-5181-4190-8066-4797cefe0189"}
    with connect(CONFIG.paths.db_path) as conn:
        for label, slot_key in slots.items():
            n_decks = conn.execute(
                "SELECT COUNT(*) FROM decks WHERE slot_key = ?", (slot_key,)).fetchone()[0]
            counts = conn.execute(
                "SELECT dc.oracle_id, COUNT(DISTINCT dc.deck_id) FROM deck_cards dc "
                "JOIN decks d ON d.deck_id = dc.deck_id WHERE d.slot_key = ? GROUP BY dc.oracle_id",
                (slot_key,)).fetchall()
            # ">=1 deck" would be vacuous (2,900 cards for Krenko); >=50% isolates the signature.
            signature = {o for o, c in counts if c >= 0.5 * n_decks}
            per_comm = cards.assign(sig=cards["oracle_id"].isin(signature)).groupby("community")["sig"].sum()
            cid = int(per_comm.idxmax())
            print(f"  {label:<7} {n_decks:5,} decks, {len(signature):3d} signature cards "
                  f"-> {per_comm.max() / len(signature):5.1%} in community {cid} "
                  f"({sizes[cid]:,} cards)")

    print("\nLARGEST COMMUNITIES  (auto-named from Scryfall tags)")
    for cid in sizes.head(12).index:
        print(f"  {cid:>3}  {sizes[cid]:>4,} cards   {community_names[cid]}")
    n_generic = sum(1 for v in community_names.values() if v.startswith("(generic"))
    print(f"  ...  {n_generic} of {len(community_names)} communities have no distinctive tag")


overview()
```

## 4. The interactive graph

ForceAtlas2 runs in the browser: edge weight becomes attraction, so tightly-linked cards pull
themselves into visible blobs. Drag, zoom, hover a node for its card name and community, and pause
the layout once it settles.

Colour is the **coarse** community (55 groups — 214 distinct colours would be indistinguishable);
position shows the fine structure. `EDGE_THRESHOLD` is the "only draw connections above X" control:
raising it thins the edges *and* drops cards left with none, which is what makes the shape readable.

```python
from ipysigma import Sigma


def display_graph(threshold: float = EDGE_THRESHOLD, *, start_layout: int = 20,
                  height: int = 800) -> Sigma:
    """Threshold the edges, drop cards left isolated, hand the rest to ipysigma."""
    H = G.edge_subgraph(
        [(u, v) for u, v, w in G.edges(data="weight") if w >= threshold]
    ).copy()
    H.remove_nodes_from([n for n, d in H.degree() if d == 0])

    for column, attr in (("name", "card"), ("community", "community"),
                         ("label", "theme"), ("coarse", "coarse"),
                         ("color_identity", "colors")):
        nx.set_node_attributes(H, cards[column].to_dict(), attr)
    nx.set_node_attributes(H, dict(H.degree(weight="weight")), "strength")

    print(f"t-score >= {threshold}: {H.number_of_nodes():,} cards, {H.number_of_edges():,} edges "
          f"({G.number_of_nodes() - H.number_of_nodes():,} dropped as isolated)")

    return Sigma(
        H, height=height,
        node_color="coarse", node_size="strength", node_size_range=(3, 20),
        node_label="card", edge_weight="weight", edge_size="weight",
        edge_size_range=(0.2, 2.0), default_edge_color="#d5d5d5",
        start_layout=start_layout, label_density=1.5, max_categorical_colors=None,
    )


display_graph()
```

---

*Optional helpers — defined, not run.*

- `show_community(7)` — a community's strongest cards as images
- `show_community("Lion Sash")` — **the community a card belongs to**, with that card marked
- `neighbors("Lion Sash")` — a card's own strongest edges, with lift alongside t-score
- `display_graph(6.0)` — redraw at a different edge threshold

```python
import difflib

from IPython.display import HTML, display

from edhcut.images import get_card_image_uri


def find_card(query: str) -> pd.Series:
    """Look a card up by name -- exact first, then case-insensitive, then fuzzy."""
    hit = cards[cards["name"].str.lower() == query.lower()]
    if len(hit):
        return hit.iloc[0]
    hit = cards[cards["name"].str.contains(query, case=False, regex=False)]
    if len(hit):
        return hit.sort_values("strength", ascending=False).iloc[0]
    close = difflib.get_close_matches(query, cards["name"].tolist(), n=3, cutoff=0.6)
    raise KeyError(f"{query!r} is not a node in this graph."
                   + (f" Did you mean: {', '.join(close)}?" if close else
                      " (It may have been dropped: no positive t-score edge survived.)"))


def show_community(which, *, limit: int = 12, width: int = 150) -> None:
    """Render a community's strongest cards as images.

    `which` is either a community id (7) or a card name ("Lion Sash"), in which case that card's
    own community is shown and the card itself is highlighted.
    """
    highlight = None
    if isinstance(which, str):
        card = find_card(which)
        which, highlight = int(card["community"]), card["name"]

    members = cards[cards["community"] == which].sort_values("strength", ascending=False)
    shown = members.head(limit)
    if highlight is not None and highlight not in set(shown["name"]):
        shown = pd.concat([members[members["name"] == highlight], shown.head(limit - 1)])

    tiles = []
    with connect(CONFIG.paths.db_path) as conn:
        for _, row in shown.iterrows():
            try:
                uri = get_card_image_uri(conn, row["name"])
            except (KeyError, ValueError):
                continue
            ring = ("box-shadow:0 0 0 4px #d64550;" if row["name"] == highlight else "")
            tiles.append(f'<img src="{uri}" width="{width}" '
                         f'title="{row["name"]} (strength {row["strength"]})" '
                         f'style="margin:3px;border-radius:9px;{ring}">')
    display(HTML(
        f"<div style='margin:14px 0 4px'><b>community {which}</b> &mdash; {len(members)} cards, "
        f"<i>{community_names[which]}</i>, modal colour "
        f"{members['color_identity'].value_counts().index[0]}"
        + (f" &mdash; showing <b>{highlight}</b>" if highlight else "") + "</div>"
        f"<div style='display:flex;flex-wrap:wrap'>{''.join(tiles)}</div>"))


def neighbors(query: str, limit: int = 15) -> pd.DataFrame:
    """A card's strongest edges. `lift` (obs/expected co-occurrence) is shown next to `t_score`
    because they disagree sharply: t-score is a significance measure, so a generic staple in
    8,000 decks scores as highly as a true synergy piece in 40. Lift is the effect size."""
    card = find_card(query)
    top = sorted(G[int(card["node"])].items(), key=lambda kv: -kv[1]["weight"])[:limit]
    others = cards.loc[[nb for nb, _ in top]]

    # `deck_count` comes from card_index, and all co-occurrences come back in one query --
    # `deck_cards` has no index on oracle_id, so a per-neighbour query means a full scan each.
    with connect(CONFIG.paths.db_path) as conn:
        total = conn.execute("SELECT COUNT(*) FROM decks").fetchone()[0]
        placeholders = ",".join("?" * len(others))
        observed = dict(conn.execute(
            f"SELECT b.oracle_id, COUNT(*) FROM deck_cards a JOIN deck_cards b "
            f"ON a.deck_id = b.deck_id WHERE a.oracle_id = ? "
            f"AND b.oracle_id IN ({placeholders}) GROUP BY b.oracle_id",
            [card["oracle_id"], *others["oracle_id"]]))

    expected = card["deck_count"] * others["deck_count"] / total
    return pd.DataFrame({
        "card": others["name"].to_numpy(),
        "t_score": [round(d["weight"], 2) for _, d in top],
        "lift": (others["oracle_id"].map(observed).fillna(0) / expected).round(2).to_numpy(),
        "decks": others["deck_count"].to_numpy(),
        "community": others["community"].to_numpy(),
        "theme": [community_names[c] for c in others["community"]],
    })
```
