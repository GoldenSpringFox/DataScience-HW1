# Embeddings exploration (task 6.2)

Two things:

1. **Finding substitutes** for three specific cards — Cultivate, Avacyn's Pilgrim, Druid of
   Purification — in both embedding spaces (`w2v`: trained on real decklists; `tfidf`: trained
   on rules text + the `struct_*` color/cost/stat features). See
   `docs/devlog/6.2-embeddings.md` for how each space is built.
2. A **2D map of the Word2Vec space** (64 dims -> 2, via t-SNE) so we can visually see which
   cards actually sit near each other, rather than only reading a ranked neighbor list.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.manifold import TSNE

from edhcut.analysis.embeddings import KB_DEV_DIR, load_embeddings, nearest_neighbors
from edhcut.config import CONFIG
from edhcut.db import connect
from edhcut.http import get_session
from edhcut.images import get_card_image
from edhcut.ingest.scryfall import normalize_name

pd.set_option("display.max_colwidth", 60)

# Keep the context-manager object itself alive (not just the yielded `conn`) -- otherwise the
# generator-based context manager can get garbage-collected mid-notebook, which runs its own
# `finally: conn.close()` and closes the connection out from under later cells.
_db_ctx = connect(CONFIG.paths.db_path)
conn = _db_ctx.__enter__()

# One shared cached/rate-limited session for every image fetch below, instead of one per call.
img_session = get_session("scryfall")

embeddings = load_embeddings(KB_DEV_DIR)
print(f"{len(embeddings):,} cards total, {embeddings['w2v_0'].notna().sum():,} with a Word2Vec vector.")

TARGET_CARDS = ["Cultivate", "Avacyn's Pilgrim", "Druid of Purification"]


def resolve_oracle_id(name: str) -> str:
    row = conn.execute(
        "SELECT oracle_id FROM card_names WHERE name_normalized = ?", (normalize_name(name),)
    ).fetchone()
    if row is None:
        raise KeyError(f"Card {name!r} not found in card_names.")
    return row[0]


TARGET_ORACLE_IDS = {name: resolve_oracle_id(name) for name in TARGET_CARDS}
TARGET_ORACLE_IDS
```

## 1. Substitutes

For each target card, top-8 neighbors in both spaces, shown as card images (not a name/id
table) with the cosine similarity captioned per card. `w2v` reflects "tends to be built
alongside"; `tfidf` reflects "does a similar thing on the card" (plus the `struct_*` color/cost/
power/toughness features).

```python
def show_card_row(rows: pd.DataFrame, title: str) -> None:
    """One row of card images (fetched live via edhcut.images.get_card_image) captioned with
    each card's cosine similarity to the query card."""
    n = len(rows)
    if n == 0:
        print(f"({title}: no neighbors found)")
        return
    fig, axes = plt.subplots(1, n, figsize=(2.4 * n, 3.8))
    if n == 1:
        axes = [axes]
    for ax, (_, row) in zip(axes, rows.iterrows()):
        image = get_card_image(conn, row["name"], session=img_session)
        ax.imshow(image)
        ax.axis("off")
        ax.set_title(f"{row['name']}\nsim {row['similarity']:.3f}", fontsize=8)
    fig.suptitle(title, fontsize=12)
    fig.tight_layout(rect=(0, 0, 1, 0.90))
    plt.show()


def show_substitutes(card_name: str, *, k: int = 8) -> None:
    oracle_id = TARGET_ORACLE_IDS[card_name]
    query_row = embeddings.loc[embeddings["oracle_id"] == oracle_id]
    show_card_row(query_row.assign(similarity=1.0), f"{card_name} (query card)")
    for space in ("w2v", "tfidf"):
        table = nearest_neighbors(embeddings, oracle_id, space=space, k=k)
        show_card_row(table, f"{card_name} — top {k} in '{space}' space")
```

```python
show_substitutes("Cultivate")
```

```python
show_substitutes("Avacyn's Pilgrim")
```

```python
show_substitutes("Druid of Purification")
```

## 2. A 2D map of the Word2Vec space

The `w2v_*` columns are 64-dimensional — not directly plottable. Reducing to 2D with t-SNE
(nonlinear, built specifically to preserve *local* neighborhood structure — the right tool for
"which cards actually sit near each other", unlike a linear projection like PCA which tends to
flatten this kind of clustering into one blob) lets us look at the whole Word2Vec vocabulary at
once.

Points are colored by color-identity bucket (colorless / mono / multicolor) as a sanity check —
if the space is capturing anything real about deck-building, mono-colored ramp/utility
creatures should visibly cluster by color even though color was never given to Word2Vec
directly (it only ever saw oracle_ids co-occurring in decks).

```python
import json

w2v_cols = [c for c in embeddings.columns if c.startswith("w2v_")]
vocab = embeddings.dropna(subset=w2v_cols).reset_index(drop=True)

# Pull color_identity for every card in one query rather than a giant `WHERE oracle_id IN (...)`
# -- 7.5k placeholders risks the SQLite bound-parameter limit for no real benefit here.
n_colors = {
    oracle_id: len(json.loads(color_identity or "[]"))
    for oracle_id, color_identity in conn.execute("SELECT oracle_id, color_identity FROM cards")
}


def bucket(oracle_id: str) -> str:
    n = n_colors.get(oracle_id, 0)
    if n == 0:
        return "Colorless"
    if n == 1:
        return "Mono"
    return "Multicolor"


vocab["color_bucket"] = vocab["oracle_id"].map(bucket)

tsne = TSNE(n_components=2, random_state=42, perplexity=30, init="pca")
xy = tsne.fit_transform(vocab[w2v_cols].to_numpy())
vocab["tsne_x"] = xy[:, 0]
vocab["tsne_y"] = xy[:, 1]

vocab["color_bucket"].value_counts()
```

```python
# Fixed categorical order/hues (never cycled) -- capped at 3 series for a scatter plot, where
# CVD-safety only validates across the full all-pairs distance matrix for the first 3 slots.
BUCKET_COLORS = {"Colorless": "#2a78d6", "Mono": "#eb6834", "Multicolor": "#1baf7a"}

fig, ax = plt.subplots(figsize=(11, 9))
for bucket_name, color in BUCKET_COLORS.items():
    subset = vocab[vocab["color_bucket"] == bucket_name]
    ax.scatter(subset["tsne_x"], subset["tsne_y"], s=8, alpha=0.45, color=color, label=bucket_name, linewidths=0)

for name, oracle_id in TARGET_ORACLE_IDS.items():
    row = vocab[vocab["oracle_id"] == oracle_id]
    if row.empty:
        continue
    x, y = row["tsne_x"].iloc[0], row["tsne_y"].iloc[0]
    ax.scatter([x], [y], s=220, marker="*", color="black", zorder=5)
    ax.annotate(name, (x, y), textcoords="offset points", xytext=(8, 8), fontsize=10, fontweight="bold")

ax.set_title(f"Word2Vec card space, t-SNE to 2D ({len(vocab):,} cards)")
ax.set_xlabel("t-SNE 1")
ax.set_ylabel("t-SNE 2")
ax.legend(loc="best", frameon=True)
plt.show()
```

### Interactive version (hover to see cards)

The static plot above is a fixed image — 8,526 points is also too dense to hover usefully even
if it were interactive. This version caps the map to the **top `TOP_N_PLAYED` most-played
cards** (real distinct-deck count from `card_index.parquet`, the same count task 6.1's
co-occurrence matrices use — not raw Word2Vec occurrence count, which triples via
`ShuffledDeckCorpus`'s 3x-per-epoch shuffling and so isn't a fair "popular card" signal), plus
the three target cards from Section 1 even if they fall outside that cap, so this view stays
comparable to the static one. It reuses the t-SNE coordinates already computed above — no
second t-SNE run.

Hovering a point fetches (or serves from the `requests-cache` cache) that card's real image
into the side panel — genuinely "look at the map, see the card," not a name/id lookup.

```python
from edhcut.analysis.cooccurrence import load_card_index

TOP_N_PLAYED = 250  # full vocab (8.5k pts) is too dense to hover usefully -- tune as needed

card_index = load_card_index(KB_DEV_DIR)[["oracle_id", "deck_count"]]
plot_df = vocab.merge(card_index, on="oracle_id", how="inner")
plot_df = plot_df.sort_values("deck_count", ascending=False).head(TOP_N_PLAYED)

# Keep the three target cards on the map even if they fall outside the top N (so this view
# stays comparable to the static one above / Section 1's neighbor lists), tagged separately so
# the plotting cell can give them their own star-marker trace.
target_ids = set(TARGET_ORACLE_IDS.values())
missing_targets = vocab[vocab["oracle_id"].isin(target_ids) & ~vocab["oracle_id"].isin(plot_df["oracle_id"])]
missing_targets = missing_targets.merge(card_index, on="oracle_id", how="left")
missing_targets["deck_count"] = missing_targets["deck_count"].fillna(0).astype(int)

plot_df = pd.concat([plot_df, missing_targets], ignore_index=True)
plot_df["is_target"] = plot_df["oracle_id"].isin(target_ids)

print(
    f"Showing {len(plot_df):,} of {len(vocab):,} w2v-vocab cards "
    f"(top {TOP_N_PLAYED} by deck count, plus {len(missing_targets)} target card(s) added back in). "
    f"Most played: {plot_df.iloc[plot_df['deck_count'].argmax()]['name']} "
    f"({plot_df['deck_count'].max():,} decks)."
)
```

```python
from io import BytesIO

from tqdm.auto import tqdm


def _png_bytes(image) -> bytes:
    buf = BytesIO()
    image.convert("RGB").save(buf, format="PNG")
    return buf.getvalue()


# Preload every thumbnail up front (rather than fetching on hover) so hovering is instant --
# `requests-cache` avoids re-downloading anything already fetched in Section 1's rows above.
thumbnails: dict[str, bytes | None] = {}
for name in tqdm(plot_df["name"], desc="Fetching card thumbnails"):
    try:
        thumbnails[name] = _png_bytes(get_card_image(conn, name, session=img_session))
    except Exception:
        thumbnails[name] = None
```

```python
import ipywidgets as widgets
import plotly.graph_objects as go
from IPython.display import display

fig = go.FigureWidget()

for bucket_name, color in BUCKET_COLORS.items():
    subset = plot_df[(plot_df["color_bucket"] == bucket_name) & ~plot_df["is_target"]]
    fig.add_scatter(
        x=subset["tsne_x"], y=subset["tsne_y"],
        mode="markers",
        marker=dict(size=8, color=color, opacity=0.7, line=dict(width=0)),
        name=bucket_name,
        customdata=subset[["name", "deck_count"]].to_numpy(),
        hovertemplate="<b>%{customdata[0]}</b><br>%{customdata[1]:,} decks<extra></extra>",
    )

# Target cards get their own trace on top: a star marker + always-visible label, mirroring the
# static plot above, regardless of which color bucket they'd otherwise fall into.
targets = plot_df[plot_df["is_target"]]
fig.add_scatter(
    x=targets["tsne_x"], y=targets["tsne_y"],
    mode="markers+text",
    marker=dict(size=16, color="black", symbol="star"),
    text=targets["name"],
    textposition="top center",
    name="Target card",
    customdata=targets[["name", "deck_count"]].to_numpy(),
    hovertemplate="<b>%{customdata[0]}</b><br>%{customdata[1]:,} decks<extra></extra>",
)

fig.update_layout(
    title=f"Word2Vec card space, t-SNE to 2D — top {len(plot_df):,} most-played cards (hover for image)",
    xaxis_title="t-SNE 1", yaxis_title="t-SNE 2",
    width=880, height=760, legend=dict(title="Color identity"),
)

image_panel = widgets.Image(value=b"", format="png", width=260)
info_panel = widgets.HTML(value="<i>Hover a point to see the card.</i>")


def _on_hover(trace, points, state):
    if not points.point_inds:
        return
    name, deck_count = trace.customdata[points.point_inds[0]]
    thumb = thumbnails.get(name)
    if thumb is not None:
        image_panel.value = thumb
    info_panel.value = f"<b>{name}</b><br>{int(deck_count):,} decks"


for trace in fig.data:
    trace.on_hover(_on_hover)

display(widgets.HBox([fig, widgets.VBox([image_panel, info_panel])]))
```

### Zoomed-in local neighborhoods

The overview above is too dense to read individual names. For each target card, zoom to its
own corner of the map and label its actual top-15 `w2v` neighbors (the same ones
`nearest_neighbors()` returned in Section 1) at their real t-SNE coordinates — a visual
cross-check that the ranked list and the map agree.

```python
vocab_xy = vocab[["oracle_id", "tsne_x", "tsne_y", "color_bucket"]]
vocab_xy_indexed = vocab_xy.set_index("oracle_id")

fig, axes = plt.subplots(1, len(TARGET_CARDS), figsize=(6.5 * len(TARGET_CARDS), 6))
for ax, (name, oracle_id) in zip(axes, TARGET_ORACLE_IDS.items()):
    if oracle_id not in vocab_xy_indexed.index:
        ax.set_title(f"{name} (no w2v vector)")
        continue
    neighbors = nearest_neighbors(embeddings, oracle_id, space="w2v", k=15)
    neighbor_xy = neighbors.merge(vocab_xy, on="oracle_id", how="left").dropna(subset=["tsne_x", "tsne_y"])

    for bucket_name, color in BUCKET_COLORS.items():
        subset = neighbor_xy[neighbor_xy["color_bucket"] == bucket_name]
        ax.scatter(subset["tsne_x"], subset["tsne_y"], s=60, color=color, label=bucket_name, zorder=3)
    for _, row in neighbor_xy.iterrows():
        ax.annotate(row["name"], (row["tsne_x"], row["tsne_y"]), fontsize=7, xytext=(5, 4), textcoords="offset points")

    tx, ty = vocab_xy_indexed.loc[oracle_id, ["tsne_x", "tsne_y"]]
    ax.scatter([tx], [ty], s=260, marker="*", color="black", zorder=5)
    ax.annotate(name, (tx, ty), fontsize=9, fontweight="bold", xytext=(6, -12), textcoords="offset points")

    ax.set_title(f"{name} — top 15 w2v neighbors")
    ax.set_xticks([])
    ax.set_yticks([])

axes[0].legend(loc="best", fontsize=8, frameon=True)
fig.tight_layout()
plt.show()
```

Observed, not spun: Cultivate's top-15 neighbors land tightly around it in the 2D map — the
visual and the ranked list agree cleanly. Avacyn's Pilgrim's and Druid of Purification's
neighbors are spread much further from their own star, several clear across the panel. This
is a real limitation of using one global t-SNE run to eyeball local neighborhoods, not a
problem with the neighbor list itself: t-SNE optimizes one 2D layout for the whole vocabulary
at once, and a card sitting in a denser or more contested region of the original 64D space
can have its true nearest neighbors pulled apart by the projection even though their cosine
similarity in the real 64D space (the number actually driving the ranked list) is unaffected.

## Notes

- t-SNE coordinates are not comparable across runs/params — only relative distances *within*
  this one embedding matter, not absolute position or axis scale.
- The zoomed insets use each target's real top-15 `w2v` neighbors (not "whatever happens to be
  spatially close after t-SNE distortion"), so they're a direct visual of the same ranking
  Section 1 already showed as images.
